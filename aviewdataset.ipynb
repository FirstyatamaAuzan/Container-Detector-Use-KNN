```python
import os
import json
import matplotlib.pyplot as plt
import numpy as np
from skimage import io
from skimage.transform import resize
from collections import defaultdict

# Path ke folder gambar dan file JSON
image_folder = 'D:/Serba Serbi Kuliah/MATERI KULIAH SMT 5/AI/UAS_AI/Container-Detector-Use-KNN/data new'
json_file = 'D:/Serba Serbi Kuliah/MATERI KULIAH SMT 5/AI/UAS_AI/Container-Detector-Use-KNN/data_new.json'

# Memuat data dari JSON dan gambar
images, labels = [], []
with open(json_file) as f:
    data = json.load(f)
    for item in data:
        image_filename = item['file_name']
        label = item['annotations']['color_label']
        image_path = os.path.join(image_folder, image_filename)
        image = io.imread(image_path)
        image_resized = resize(image, (64, 64))
        images.append(image_resized.flatten().tolist())
        labels.append(label)

# Dictionary untuk menyimpan gambar berdasarkan label
label_images = defaultdict(list)

# Mengelompokkan gambar berdasarkan label
for i in range(len(labels)):
    if len(label_images[labels[i]]) < 5:  # Ambil maksimal 5 gambar per label
        image_reshaped = np.array(images[i]).reshape(64, 64, -1)  # Mengembalikan ke ukuran asli (64x64x3 untuk RGB)
        label_images[labels[i]].append(image_reshaped)

# Menampilkan 5 gambar untuk setiap label
for label, image_list in label_images.items():
    plt.figure(figsize=(15, 5))
    plt.suptitle(f"Label: {label}", fontsize=16)
    for i, img in enumerate(image_list):
        plt.subplot(1, 5, i + 1)
        plt.imshow(img.astype('float32'))  # Pastikan tipe data cocok untuk visualisasi
        plt.axis('off')
    plt.show()
```
