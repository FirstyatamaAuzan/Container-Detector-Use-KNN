```python
import os
import json
from collections import defaultdict
import random
from skimage import io
from skimage.transform import resize

image_folder = 'D:/Serba Serbi Kuliah/MATERI KULIAH SMT 5/AI/UAS_AI/Container-Detector-Use-KNN/data new'
json_file = 'D:/Serba Serbi Kuliah/MATERI KULIAH SMT 5/AI/UAS_AI/Container-Detector-Use-KNN/data_new.json'
```

```python
def load_data_from_json_and_images(image_folder, json_file):
    images = []
    labels = []
    filenames = [] 
    
    # Memuat JSON
    with open(json_file) as f:
        data = json.load(f)  # JSON Anda adalah list, bukan dictionary

    # Membaca gambar dan label dari JSON
    for item in data:
        image_filename = item['file_name']  # Mengambil nama file gambar
        label = item['annotations']['color_label']  # Mengambil label warna dari anotasi
        
        # Memuat gambar dari folder
        image_path = os.path.join(image_folder, image_filename)
        image = io.imread(image_path)
        
        # Ubah ukuran gambar jika perlu (agar ukuran konsisten)
        image_resized = resize(image, (64, 64))  
        
        # Flatten gambar menjadi array 1D dengan list
        image_flattened = image_resized.flatten().tolist()  
        
        images.append(image_flattened)
        labels.append(label)
        filenames.append(image_filename) 
    
    return images, labels, filenames
```

```python
def euclidean_distance(vec1, vec2):
    sum_squared = sum((a - b) ** 2 for a, b in zip(vec1, vec2))
    return sum_squared ** 0.5
```

```python
def knn_classify(test_data, train_data, train_labels, k):
    distances = []
    for i in range(len(train_data)):
        dist = euclidean_distance(test_data, train_data[i])
        distances.append((dist, train_labels[i]))
    
    distances.sort(key=lambda x: x[0])
    k_nearest = distances[:k]
    
    label_votes = {}
    for _, label in k_nearest:
        if label in label_votes:
            label_votes[label] += 1
        else:
            label_votes[label] = 1
    
    return max(label_votes, key=label_votes.get)
```

```python
# Memuat data
images, labels, filenames = load_data_from_json_and_images(image_folder, json_file)

# Stratification
label_to_indices = defaultdict(list)
for i, label in enumerate(labels):
    label_to_indices[label].append(i)

# Membagi data menjadi train dan test dengan stratification
train_indices, test_indices = [], []
for label, indices in label_to_indices.items():
    random.shuffle(indices)  # Acak data per label
    split_point = int(0.8 * len(indices))  # 80% untuk train, 20% untuk test
    train_indices.extend(indices[:split_point])
    test_indices.extend(indices[split_point:])

# Membuat data train dan test berdasarkan indeks
train_data = [images[i] for i in train_indices]
train_labels = [labels[i] for i in train_indices]
train_filenames = [filenames[i] for i in train_indices]

test_data = [images[i] for i in test_indices]
test_labels = [labels[i] for i in test_indices]
test_filenames = [filenames[i] for i in test_indices]
```

```python
from collections import defaultdict

# KNN Classification
k = 1  # Tetangga terdekat
correct_predictions = 0

# Confusion matrix initialization
unique_labels = sorted(set(test_labels))  # Pastikan semua label unik ditemukan
confusion_matrix = {label: {inner_label: 0 for inner_label in unique_labels} for label in unique_labels}

# Iterasi setiap sampel data uji
for i, test_sample in enumerate(test_data):
    predicted_label = knn_classify(test_sample, train_data, train_labels, k)
    true_label = test_labels[i]

    # Update confusion matrix
    confusion_matrix[true_label][predicted_label] += 1

# Akurasi keseluruhan menggunakan confusion matrix
tp_sum = sum(confusion_matrix[label][label] for label in unique_labels)  # Total True Positives
total_predictions = sum(
    sum(confusion_matrix[label].values()) for label in unique_labels
)  # Total Semua Prediksi
overall_accuracy = (tp_sum / total_predictions) * 100

# Print akurasi keseluruhan
print(f"\nOverall Accuracy (from Confusion Matrix): {overall_accuracy:.2f}%")

# Confusion Matrix
print("\nConfusion Matrix:")
print("    " + "  ".join(map(str, unique_labels)))
for label in unique_labels:
    row = [confusion_matrix[label][inner_label] for inner_label in unique_labels]
    print(f"{label} {'  '.join(map(str, row))}")

# Precision, Recall, F1-Score untuk semua label dan macro-averaged
print("\nMetrics for All Labels:")
precision_list = []
recall_list = []
f1_score_list = []

for label in unique_labels:
    tp = confusion_matrix[label][label]  # True Positives
    fp = sum(confusion_matrix[other][label] for other in unique_labels if other != label)  # False Positives
    fn = sum(confusion_matrix[label][other] for other in unique_labels if other != label)  # False Negatives

    # Hindari pembagian dengan nol
    precision = tp / (tp + fp) if tp + fp > 0 else 0
    recall = tp / (tp + fn) if tp + fn > 0 else 0
    f1_score = (2 * precision * recall) / (precision + recall) if precision + recall > 0 else 0

    # Simpan nilai untuk macro-averaging
    precision_list.append(precision)
    recall_list.append(recall)
    f1_score_list.append(f1_score)

    print(f"{label.capitalize()}:")
    print(f"  Precision: {precision:.2f}")
    print(f"  Recall:    {recall:.2f}")
    print(f"  F1-Score:  {f1_score:.2f}")

# Macro-Averaged Metrics
macro_precision = sum(precision_list) / len(precision_list)
macro_recall = sum(recall_list) / len(recall_list)
macro_f1_score = sum(f1_score_list) / len(f1_score_list)

print("\nMacro-Averaged Metrics:")
print(f"  Macro Precision: {macro_precision:.2f}")
print(f"  Macro Recall:    {macro_recall:.2f}")
print(f"  Macro F1-Score:  {macro_f1_score:.2f}")
```
